```python
#A)

def percentageLate(zone, time_block, promise):
  late = 0
  total = 0

#If zone is all, check all zones, otherwise only look at the one that was entered
  if zone == "all":
    zoneToCheck = ZONES
  else:
    zoneToCheck = [zone]

#If timeblock is all, check all timeblocks, otherwise only look at the one that was entered
  if time_block == "all":
    timeblockToCheck = TIME_BLOCKS
  else:
    timeblockToCheck = [time_block]

#Check the zones to check and timeblock, and call the delivery_times function to get the times
  for z in zoneToCheck:
    for t in timeblockToCheck:
      time = delivery_times(z, t, promise, 42)

#Sum the times, and the times that it is late
      total += len(time)
      late += np.sum(time > promise)

#Return late tiems as a %
  return round(late/total * 100)

#Show scenarios
print("Scenario 1, zone and time block provided: ", percentageLate("Central", "Fri/Sat eve", 45), "%")

print("Scenario 2, zone is all and time block is provided: ", percentageLate("all", "Lunch", 45), "%")

print("Scenario 3, zone is provided and time block is all: ", percentageLate("North", "all", 45), "%")

print("Scenario 4, zone and time block are all: ", percentageLate("all", "all", 45), "%")
```

```python
#B)


allZonesTimes = []

#Run a loop to get delivery times for all zone and timeblock combinations
for z in ZONES:
  for t in TIME_BLOCKS:
#Call the delivery times function and sum all the times there is a late delivery
      time = delivery_times(z, t, 45, 42)
      lateRate = np.sum(time > 45) / len(time) * 100
#Add all late results to the array
      allZonesTimes.append((z, t, lateRate))
#Sort late results by ascending order
allZonesTimes.sort(key=lambda x: x[2], reverse=True)
#iterate through array and print all results
for z, t, lateRate in allZonesTimes:
  print(f"Zone: {z}, Time Block: {t}. Late Percentage: {lateRate:.2f}%")
```

```python
#C), was able to resuse the same function created in part A, just modify the calculation

def averageDeliveryTime(zone, time_block, promise):
  minutesTotal = 0
  total = 0

#If zone is all, check all zones, otherwise only look at the one that was entered
  if zone == "all":
    zoneToCheck = ZONES
  else:
    zoneToCheck = [zone]

#If timeblock is all, check all timeblocks, otherwise only look at the one that was entered
  if time_block == "all":
    timeblockToCheck = TIME_BLOCKS
  else:
    timeblockToCheck = [time_block]

#Check the zones to check and timeblock, and call the delivery_times function to get the times
  for z in zoneToCheck:
    for t in timeblockToCheck:
      time = delivery_times(z, t, promise, 42)

#Sum the times, and the total minutes
      total += len(time)
      minutesTotal += np.sum(time)

#Return average minutes
  return round(minutesTotal/total)

print("Scenario 1, zone and time block provided: ", averageDeliveryTime("Central", "Fri/Sat eve", 45), "min")

print("Scenario 2, zone is all and time block is provided: ", averageDeliveryTime("all", "Lunch", 45), "min")

print("Scenario 3, zone is provided and time block is all: ", averageDeliveryTime("North", "all", 45), "min")

print("Scenario 4, zone and time block are all: ", averageDeliveryTime("all", "all", 45), "min")
```

```python
#A

#Define new function to determine late cost per order
def costPerLateOrder(costs):
  #Calculate late cost by pulling data from dictionary
  lateCost = costs['refund'] + costs['margin']*costs['churn_orders']
  return lateCost

print(costPerLateOrder(COSTS))
```

Part B

I believe that the best way to choose the range would be to check in small intervals. This way we can see at which point do we reach the maximum, and can have a good degree of granularity within the readings. The range I will go with will be by checking in 3 minute intervals, and going to 60 minutes.

This range can effectively capture the low end and higher end while being able to locate the optimal point.

```python
#B

promiseTimes = [3, 6, 9, 12, 15, 18, 21, 24, 27, 30, 33, 36, 39, 42, 45, 48, 51, 54, 57, 60]

def costPerPromiseTime(zone, time_block, promiseTimes, costs):

  bestProfit = 0
  bestPromiseTime = 0
  bestZone = 0
  bestTimeBlock = 0
    #If zone is all, check all zones, otherwise only look at the one that was entered
  if zone == "all":
    zoneToCheck = ZONES
  else:
    zoneToCheck = [zone]

#If timeblock is all, check all timeblocks, otherwise only look at the one that was entered
  if time_block == "all":
    timeblockToCheck = TIME_BLOCKS
  else:
    timeblockToCheck = [time_block]

#Check the zones to check and timeblock, and call the delivery_times function to get the times
  for z in zoneToCheck:
    for t in timeblockToCheck:
      for p in promiseTimes:
        late = 0
        total = 0
        time = delivery_times(z, t, p, 42)
        total += len(time)
        late += np.sum(time > p)
        totalProfit = total*costs['margin'] - late*costPerLateOrder(costs)
        if totalProfit > bestProfit:
          bestProfit = totalProfit
          bestPromiseTime = p
          bestZone = z
          bestTimeBlock = t

  return print("For the", bestZone, "zone, and the Time Block", bestTimeBlock, ", the Best profit is: $", round(bestProfit, 2), ", and the optimal promise time is:", bestPromiseTime, "min.")

costPerPromiseTime("Central", "Fri/Sat eve", promiseTimes, COSTS)


```
